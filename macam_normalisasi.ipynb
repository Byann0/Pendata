# Normalisasi Data

**Normalisasi** adalah proses penskalaan nilai atribut dari sebuah dataset sehingga berada dalam rentang nilai yang seragam.

Dalam konteks penambangan data adalah bagaimana mendapatkan informasi yang berharga untuk pijakan pengambilan keputusan atau tujuan tertentu.

Terdapat 3 cara untuk melakukan normalisasi data sebagai berikut :

```python
import pandas as pd
import numpy as np

df = pd.read_csv('Dataset Pelatihan.csv')

df_sample = df[['UTS']].iloc[0:5].copy()

print("Dataset Contoh :")
print(df_sample)
```

## 1. Min-Max Scaling

Metode ini melakukan transformasi linear pada data asli. Data kalian akan "diperas" masuk ke dalam rentang baru, paling sering adalah 0 sampai 1.

Rumus:

$$x' = \frac{x - \text{min}(x)}{\text{max}(x) - \text{min}(x)}$$

- Sangat sensitif terhadap outlier. Jika ada satu orang punya nilai UTS 1000 (salah input), maka teman-teman yang nilainya 80-90 akan terlihat sangat kecil (mendekati 0).

Contoh:

| No. | IPK |   PO  | JML |
|:---:|:---:|:-----:|:---:|
|  1. |  2  |2000000|  2  |
|  2. |  3  |3000000|  3  |
|  3. |  4  |2000000|  2  |
|  4. |  2  |2000000|  3  |
|  5. |  3  |3000000|  2  |
|  6. |  4  |4000000|  3  |
|  7. |  2  |3000000|  ?  |

- Perhitungan manual pada kolom IPK objek kedua

$$
x' = \frac{3 - 2}{4 - 2} = \frac{1}{2} = 0.5
$$

- Perhitungan manual pada kolom PO objek kedua

$$
x' = \frac{3000000 - 2000000}{4000000 - 2000000} = \frac{1000000}{2000000} = 0.5
$$

- Perhitungan manual pada kolom JML objek kedua

$$
x' = \frac{3 - 2}{3 - 2} = \frac{1}{1} = 1
$$

Setelah selesai akan didapat:

| No. | IPK |   PO  | JML |
|:---:|:---:|:-----:|:---:|
|  1. |  0  |   0   |  0  |
|  2. | 0.5 |  0.5  |  1  |
|  3. |  1  |   0   |  0  |
|  4. |  0  |   0   |  1  |
|  5. | 0.5 |  0.5  |  0  |
|  6. |  1  |   1   |  1  |
|  7. |  0  |  0.5  |  ?  |

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

data = {
    'No': [1, 2, 3, 4, 5, 6],
    'IPK': [2, 3, 2, 3, 2, 3],
    'PO': [2000000, 3000000, 2000000, 2000000, 3000000, 4000000],
    'JML': [2, 3, 2, 3, 2, 3]
}
df = pd.DataFrame(data)
kolom = ['IPK', 'PO', 'JML']
data_normalize = pd.DataFrame(data)
for col in kolom:
    min_val = df[col].min()
    max_val = df[col].max()
    
    data_normalize[col] = (df[col] - min_val) / (max_val - min_val)

print("Data Sebelum Dinormalisasi")
print(df)
print()
print("Data Setelah Dinormalisasi")
print(data_normalize)
```

## 2. Z-Score Standardization

Metode ini tidak peduli pada batas bawah atau atas. Dia fokus pada distribusi data. Hasilnya adalah data yang memiliki rata-rata ($\mu$) = 0 dan Standar Deviasi ($\sigma$) = 1.

Rumus Z-score:

$$z = \frac{x - \mu}{\sigma}$$

- Lebih tahan banting terhadap outlier dibanding Min-Max. Jika data kalian mengikuti distribusi normal (Gausian), metode ini wajib dipilih.

Rumus Standar Deviasi:

$$
\sigma = \sqrt{\frac{1}{N}\sum_{i=1}^{N}(x_i-\mu)^2}
$$

Contoh:

| No. | IPK |   PO  | JML |
|:---:|:---:|:-----:|:---:|
|  1. |  2  |2000000|  2  |
|  2. |  3  |3000000|  3  |
|  3. |  4  |2000000|  2  |
|  4. |  2  |2000000|  3  |
|  5. |  3  |3000000|  2  |
|  6. |  4  |4000000|  3  |

Perhitungan Standar Deviasi pada kolom IPK

Rumus Umum:

$$\sigma = \sqrt{\frac{1}{N} \sum_{i=1}^{N} (x_i - \mu)^2}$$

Langkah perhitungan berdasarkan data:

- $$\sigma = \sqrt{\frac{1}{6} ((2-3)^2 + (3-3)^2 + (4-3)^2 + (2-3)^2 + (3-3)^2 + (4-3)^2)}$$
- $$\sigma = \sqrt{\frac{1}{6} (1 + 0 + 1 + 1 + 0 + 1)}$$
- $$\sigma = \sqrt{\frac{1}{6} (4)} = \sqrt{0.666667} = 0.816497$$

```python
from sklearn.preprocessing import StandardScaler
import numpy as np

data = {
    'No': [1, 2, 3, 4, 5, 6],
    'IPK': [2, 3, 4, 2, 3, 4],
    'PO': [2000000, 3000000, 2000000, 2000000, 3000000, 4000000],
    'JML': [2, 3, 2, 3, 2, 3]
}

ipk_data = np.array(data['IPK'])
std_populasi = np.std(ipk_data)
print(f"Standar Deviasi Populasi: {std_populasi}")
```

Perhitungan Z-Score Normalization

Rumus Umum:

$$z = \frac{x - \mu}{\sigma}$$

Menggunakan nilai $\mu = 3$ dan $\sigma = 0.816497$, berikut adalah hasil transformasinya:

- z1:$$z1 = \frac{2 - 3}{0.816497} = \frac{-1}{0.816497} = -1.22474$$
- z2:$$z2 = \frac{3 - 3}{0.816497} = \frac{0}{0.816497} = 0$$
- z3:$$z3 = \frac{4 - 3}{0.816497} = \frac{1}{0.816497} = 1.224745$$
- z4:$$z4 = \frac{2 - 3}{0.816497} = \frac{-1}{0.816497} = -1.22474$$
- z5:$$z5 = \frac{3 - 3}{0.816497} = \frac{0}{0.816497} = 0$$
- z6:$$z6 = \frac{4 - 3}{0.816497} = \frac{1}{0.816497} = 1.224745$$

```python
mean_ipk = np.mean(ipk_data)

z_scores = (ipk_data - mean_ipk) / std_populasi

print(f"Rata-rata IPK: {mean_ipk}")
print(f"Standar Deviasi: {std_populasi:.4f}")
print("-" * 30)
print("No | IPK | Z-Score")
for i in range(len(data['No'])):
    print(f"{data['No'][i]}  | {data['IPK'][i]}   | {z_scores[i]:.4f}")
```

## 3. Decimal Scaling

Metode ini menormalkan data dengan cara menggerakkan nilai desimal dari data tersebut. Jumlah pergerakan desimalnya tergantung pada nilai absolut maksimum dari data tersebut.

Rumus:

$$v' = \frac{v}{10^j}$$

(Di mana $j$ adalah bilangan bulat terkecil yang membuat nilai absolut maksimum dari $v' < 1$)

- Sangat jarang dipakai di kompetisi data science modern, tapi sering muncul di ujian Teori Data Mining. Ini menjaga agar angka tetap terlihat seperti "versi kecil" dari aslinya.

Contoh:

| No. | IPK |   PO  | JML |
|:---:|:---:|:-----:|:---:|
|  1. |  2  |2000000|  2  |
|  2. |  3  |3000000|  3  |
|  3. |  4  |2000000|  2  |
|  4. |  2  |2000000|  3  |
|  5. |  3  |3000000|  2  |
|  6. |  4  |4000000|  3  |

Nilai absolut terbesar pada kolom PO adalah 4000000 yang memiliki 7 digit angka sehingga, $j = 7$.

Perhitungan Decimal Scaling pada kolom PO

Rumus umum:

$$v' = \frac{v}{10^j}$$

Diketahui nilai pembagi:

$$10^7 = 10.000.000$$

Langkah-langkah perhitungan:

- $$v1 = \frac{2.000.000}{10.000.000} = 0.2$$
- $$v2 = \frac{3.000.000}{10.000.000} = 0.3$$
- $$v3 = \frac{2.000.000}{10.000.000} = 0.2$$
- $$v4 = \frac{2.000.000}{10.000.000} = 0.2$$
- $$v5 = \frac{3.000.000}{10.000.000} = 0.3$$
- $$v6 = \frac{4.000.000}{10.000.000} = 0.4$$

```python
import numpy as np
from sklearn.preprocessing import FunctionTransformer

data = {
    'No': [1, 2, 3, 4, 5, 6],
    'IPK': [2, 3, 4, 2, 3, 4],
    'PO': [2000000, 3000000, 2000000, 2000000, 3000000, 4000000],
    'JML': [2, 3, 2, 3, 2, 3]
}

po_array = np.array(data['PO'])

max_val = np.max(np.abs(po_array))

j = len(str(int(max_val)))

po_scaled = po_array / (10**j)

print(f"Data Asli PO: {data['PO']}")
print(f"Nilai j: {j}")
print(f"Hasil Decimal Scaling: {po_scaled}")
```
